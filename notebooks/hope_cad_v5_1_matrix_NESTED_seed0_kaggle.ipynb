```python
PHASE='7.2'
JOB='NESTED_seed0'
SESSION_HOURS=9.0
SOURCE_OVERRIDE=None
PARENT_OVERRIDE=None
MVTEC_OVERRIDE=None
CHECKPOINT_OVERRIDE=None

import os,sys,time,json,hashlib,zipfile,shutil
from pathlib import Path,PurePosixPath
STARTED=time.monotonic()
os.environ['PYTHONDONTWRITEBYTECODE']='1'
os.environ.setdefault('CUBLAS_WORKSPACE_CONFIG',':4096:8')
sys.dont_write_bytecode=True
INPUT=Path('/kaggle/input');ROOT=Path('/tmp/hope_matrix_7_2_source_r1')
WORK=Path('/kaggle/working/hope_matrix_NESTED_seed0')
ASSETS=Path('/tmp/hope_matrix_assets_NESTED_seed0')
OUTPUT=Path('/kaggle/working/hope_cad_matrix_NESTED_seed0_evidence.zip')
print('[MATRIX BOOTSTRAP] finding hash-bound source',flush=True)
def safe(root,name):
    p=PurePosixPath(name)
    if p.is_absolute() or p.as_posix()!=name or '..' in p.parts or '\\' in name or ':' in name:raise ValueError('Unsafe source path')
    dest=root/name
    if not dest.resolve().is_relative_to(root.resolve()):raise ValueError('Source path escapes target')
    return dest
def source_manifest(p):
    if p.is_dir():return json.loads((p/'matrix_manifest.json').read_text())
    with zipfile.ZipFile(p) as z:
        ns=z.namelist()
        if len(ns)!=len(set(ns)):raise ValueError('Duplicate source entries')
        m=json.loads(z.read('matrix_manifest.json'))
        if set(ns)!=set(m['files'])|{'matrix_manifest.json'}:raise ValueError('Source inventory changed')
        return m
choices=[]
paths=[Path(SOURCE_OVERRIDE)] if SOURCE_OVERRIDE else [p.parent for p in INPUT.rglob('matrix_manifest.json')]+list(INPUT.rglob('*source*.zip'))
for p in paths:
    try:
        m=source_manifest(p)
        if m.get('schema')=='hope_support_v5_1_matrix_7_2_source_v1':choices.append((p,m))
    except (KeyError,zipfile.BadZipFile):
        if SOURCE_OVERRIDE:raise
if not choices or len({json.dumps(m,sort_keys=True) for p,m in choices})!=1:raise ValueError('Expected ONE matrix source identity; set SOURCE_OVERRIDE')
src,m=sorted(choices,key=lambda x:str(x[0]))[0]
if ROOT.exists():raise ValueError('Source root already exists; start a fresh kernel/session')
ROOT.mkdir(parents=True)
archive=zipfile.ZipFile(src) if src.is_file() else None
try:
    for name,h in m['files'].items():
        original=safe(src,name) if not archive else None
        if original and original.is_symlink():raise ValueError('Source symlink')
        raw=archive.read(name) if archive else original.read_bytes()
        if hashlib.sha256(raw).hexdigest()!=h:raise ValueError('Source SHA mismatch: '+name)
        dst=safe(ROOT,name);dst.parent.mkdir(parents=True,exist_ok=True);dst.write_bytes(raw)
finally:
    if archive:archive.close()
(ROOT/'matrix_manifest.json').write_text(json.dumps(m,sort_keys=True))
sys.path.insert(0,str(ROOT));os.chdir(ROOT)
print('[MATRIX BOOTSTRAP] verified',len(m['files']),'files; frozen core plus readonly matrix orchestration',flush=True)
from notebook_support.matrix_v5_1.common import verify_source
ORCHESTRATION,REFERENCE,CORE_SOURCE=verify_source()
print('[MATRIX IDENTITY]',json.dumps(ORCHESTRATION),flush=True)
```
